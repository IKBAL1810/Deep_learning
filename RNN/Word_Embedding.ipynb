```python
# Word Embedding Techniques using Embedding Layer in PyTorch

import torch
import torch.nn as nn
from torch.nn.utils.rnn import pad_sequence

```

```python
# ---------------------------------------
# Input Sentences
# ---------------------------------------
sentences = [
    'the glass of milk',
    'the glass of juice',
    'the cup of tea',
    'I am a good boy',
    'I am a good developer',
    'understand the meaning of words',
    'your videos are good'
]
```

```python

# ---------------------------------------
# Tokenizer (Manual Vocabulary Builder)
# ---------------------------------------
def build_vocab(sent_list):
    vocab = {"<PAD>": 0}
    idx = 1
    for sentence in sent_list:
        for word in sentence.lower().split():
            if word not in vocab:
                vocab[word] = idx
                idx += 1
    return vocab

vocab = build_vocab(sentences)
voc_size = len(vocab)
print("Vocabulary:", vocab)
```

```python

# ---------------------------------------
# Convert sentences → integer encoded
# ---------------------------------------
encoded_sentences = []
for s in sentences:
    encoded = [vocab[word] for word in s.lower().split()]
    encoded_sentences.append(torch.tensor(encoded))

print("\nInteger Encoded Sentences:")
for e in encoded_sentences:
    print(e)

# ---------------------------------------
# Padding using PyTorch
# ---------------------------------------
padded_sequences = pad_sequence(encoded_sentences, batch_first=True, padding_value=0)
print("\nPadded Sequences:")
print(padded_sequences)

# Length of sequence
sent_length = padded_sequences.size(1)

# ---------------------------------------
# PyTorch Embedding Layer
# ---------------------------------------
embedding_dim = 10

embedding = nn.Embedding(num_embeddings=voc_size, embedding_dim=embedding_dim)

# ---------------------------------------
# Get Embeddings
# ---------------------------------------
embedded_output = embedding(padded_sequences)

print("\nEmbedding Output Shape:", embedded_output.shape)
print("\nEmbedding Output for All Sentences:")
print(embedded_output)

print("\nEmbedding for First Sentence:")
print(embedded_output[0])
```
